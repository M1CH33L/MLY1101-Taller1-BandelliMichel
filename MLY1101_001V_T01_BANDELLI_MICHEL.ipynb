## Consumir la API de KrakenKeys: Búsqueda de Juegos

```python
# Importa la biblioteca 'requests' para hacer peticiones HTTP
import requests
import json

KRAKENKEYS_API_KEY = "5|ADJtds2zEzfPJmuinii96ChLm9BPTGDhTR3Se9My13fcebd8" # Mi clave :)
print("Clave de API cargada con éxito.")
```

```python
# Define las cabeceras para la autenticación
headers = {
    'Authorization': f'Bearer {KRAKENKEYS_API_KEY}'
}

print("Cabeceras de autenticación configuradas.")
```

```python
# Define la URL base de la API
BASE_URL = 'https://krakenkeys.com/api/v1/developer'

# Define el endpoint específico para buscar juegos
SEARCH_GAMES_ENDPOINT = f'{BASE_URL}/games/search'

# Define el nombre del juego que quieres buscar
game_name = 'Kingdom Hearts'

# Define los parámetros de la consulta
params = {
    'query': game_name
}

print(f"Buscando el juego: '{game_name}' en el endpoint: {SEARCH_GAMES_ENDPOINT}")
```

```python
import pandas as pd

# Lista de nombres de juegos para buscar
game_names_to_process = [
    'Kingdom Hearts',
    'Final Fantasy',
    'Zelda',
    'Mario',
    'God of war',
    'Grand Theft Auto',
    'Need For Speed',
    'Age of Empire',
    'Civilization',
    'Fallout',
    'Elder Scrolls',
    'Xenoblade'
]

basic_games_data = [] # Para almacenar la información básica

print("Iniciando la búsqueda de información básica y obtención de IDs...")

# --- Loop 1: Obtener información básica y IGDB IDs (hasta 20 resultados por búsqueda) ---
for game_name in game_names_to_process:
    #print(f"\nBuscando: {game_name}")
    params_search = {'query': game_name}
    try:
        response_search = requests.get(SEARCH_GAMES_ENDPOINT, headers=headers, params=params_search)
        response_search.raise_for_status() # Lanza HTTPError para errores 4xx/5xx
        search_results = response_search.json()

        if search_results and 'data' in search_results and len(search_results['data']) > 0:
            #print(f"  - Encontrados {len(search_results['data'])} resultados para '{game_name}'.")
            for game_item in search_results['data']:
                basic_games_data.append({
                    'search_query_original': game_name, # Nombre original de la búsqueda
                    'igdb_id': game_item.get('igdb_id'),
                    'name': game_item.get('name'),
                    'slug': game_item.get('slug'),
                    'image_url': game_item.get('image_url')
                })
        else:
            #print(f"  - No se encontraron resultados para '{game_name}'.")
            basic_games_data.append({
                'search_query_original': game_name,
                'igdb_id': None,
                'name': None,
                'slug': None,
                'image_url': None
            })

    except requests.exceptions.RequestException as e:
        print(f"  - Error al buscar '{game_name}': {e}")
        basic_games_data.append({
            'search_query_original': game_name,
            'igdb_id': None,
            'name': None,
            'slug': None,
            'image_url': None
        })

# Convertir la información básica a DataFrame
df_basic_info = pd.DataFrame(basic_games_data)
print("\n--- Información Básica de Juegos (Primeras 5 filas) ---")
display(df_basic_info.head())
```

## Consumir la API de RAWG: Búsqueda de Publishers

```python
# Define el endpoint específico para buscar publishers en RAWG
RAWG_PUBLISHERS_ENDPOINT = f'{RAWG_BASE_URL}/publishers'

publishers_data = [] # Para almacenar la información de los publishers

print("Iniciando la búsqueda de publishers usando la API de RAWG...")

# --- Loop para obtener información de publishers usando la API de RAWG ---
page_number = 1
max_pages = 5 # Obtener 5 páginas para asegurar > 200 publishers

while len(publishers_data) < 200 and page_number <= max_pages:
    params_search_publishers = {
        'key': RAWG_API_KEY,
        'page': page_number,
        'page_size': 40 # Obtener 40 publishers por página
    }
    try:
        response_search_publishers = requests.get(RAWG_PUBLISHERS_ENDPOINT, params=params_search_publishers)
        response_search_publishers.raise_for_status() # Lanza HTTPError para errores 4xx/5xx
        search_results_publishers = response_search_publishers.json()

        if search_results_publishers and 'results' in search_results_publishers and len(search_results_publishers['results']) > 0:
          # Ejecuta si en cada busqueda hay resultados y si esos resultados son mas de 0
            for publisher_item in search_results_publishers['results']:
                publishers_data.append({
                    'publisher_id': publisher_item.get('id'),
                    'name': publisher_item.get('name'),
                    'slug': publisher_item.get('slug'),
                    'games_count': publisher_item.get('games_count'),
                    'image_background': publisher_item.get('image_background')
                })
            page_number += 1
        else:
            print(f"  - No se encontraron más publishers en la página {page_number}.")
            break # Salir del bucle si no hay más resultados

    except requests.exceptions.RequestException as e:
        print(f"  - Error al buscar publishers en la página {page_number}: {e}")
        break # Salir del bucle en caso de error

print(f"Búsqueda de publishers en RAWG finalizada. Se obtuvieron {len(publishers_data)} entradas.")
```

```python
# Convertir la información de publishers de RAWG a DataFrame
df_rawg_publishers = pd.DataFrame(publishers_data)

print("\n--- Información de Publishers de RAWG (Primeras 5 filas) ---")
display(df_rawg_publishers.sample(5))

print("\n--- Dimensiones del DataFrame de Publishers de RAWG ---")
print(f"Forma dataset: {df_rawg_publishers.shape}")
```

```python
try:
  df_rawg_publishers.to_csv(
        "Publishers_rawg_api.csv",   # Nombre del archivo
        index=False,      # No incluir la columna de índice
        sep=",",          # Separador (por defecto es coma)
        encoding="utf-8", # Codificación recomendada
        na_rep="N/A"      # Representación para valores faltantes
  )
  print("DataFrame de Publishers de RAWG descargado como CSV satisfactoriamente.")
  print("En caso de estar ejecutando en Colab, debe descargar manualmente desde el icono de la carpeta -> Publishers_rawg_api.csv")
except Exception as e:
  print("ERROR EN LA DESCARGA DEL CSV de Publishers de RAWG")
  print("ERROR EXACTO:\n",e)
```
